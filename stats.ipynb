```python
import json
import numpy as np

with open("c:/Users/arshm/Downloads/xlrs_first_2k_captions.json", "r") as f:
    data = json.load(f)

caption_lengths_char = [len(item["description"]["caption"]) for item in data]
caption_lengths_words = [len(item["description"]["caption"].split()) for item in data]

mean_length = np.mean(caption_lengths_char)
variance_length = np.var(caption_lengths_char)

mean_length_1 = np.mean(caption_lengths_words)
variance_length_1 = np.var(caption_lengths_words)

print("mean and std of characters", mean_length, np.sqrt(variance_length))
print("mean and std of words", mean_length_1, np.sqrt(variance_length_1))
```
