```python
from dataclasses import dataclass
from typing import Optional

@dataclass
class VLMConfig:
    model_id: str
    projector_layer_name: str
    image_token_str: str
    vision_start_token: Optional[str] = None
    vision_end_token: Optional[str] = None
    use_causal_lm: bool = False

# To add a new model, inspect it with print(model) and find the layer
# that projects visual features to the language dimension.

SUPPORTED_MODELS = {

    # Uses a simple <image> token which is expanded into 576 embeddings.
    "llava-1.5": VLMConfig(
        model_id="llava-hf/llava-1.5-7b-hf",
        projector_layer_name="multi_modal_projector",
        image_token_str="<image>",
        use_causal_lm=False
    ),

    "qwen2.5-vl-3b": VLMConfig(
        model_id="Qwen/Qwen2.5-VL-3B-Instruct",
        projector_layer_name="visual.merger",
        image_token_str="<|vision_start|>",
        vision_end_token="<|vision_end|>",
        use_causal_lm=False
    ),

    # verify layer name using print(model) for any new models which need to be added
    "kimi-vl": VLMConfig(
        model_id="moonshotai/Kimi-VL-A3B-Instruct", 
        projector_layer_name="model.model.vision_projection", 
        image_token_str="<image>",
        use_causal_lm=True 
)
}
```

```python
import torch
from contextlib import contextmanager


class VLMTracker:
    def __init__(self, model, config):
        self.model = model
        self.config = config
        self.visual_token_count = 0
        self.hook_handle = None
        self._validate_config()

    def _validate_config(self):
        try:
            self.model.get_submodule(self.config.projector_layer_name)
        except AttributeError:
            pass

    def _hook_fn(self, module, input, output):
        if isinstance(output, tuple):
            output_tensor = output[0]
        else:
            output_tensor = output
        self.visual_token_count = output_tensor.shape[1]

    @contextmanager
    def track(self):
        try:
            target_module = self.model.get_submodule(self.config.projector_layer_name)
        except AttributeError:
            if "." in self.config.projector_layer_name:
                parts = self.config.projector_layer_name.split('.')
                curr = self.model
                for part in parts:
                    curr = getattr(curr, part)
                target_module = curr
            else:
                target_module = getattr(self.model, self.config.projector_layer_name)
                
        self.hook_handle = target_module.register_forward_hook(self._hook_fn)
        try:
            yield self
        finally:
            if self.hook_handle:
                self.hook_handle.remove()
                self.hook_handle = None

    def split_embeddings(self, final_hidden_states, input_ids, tokenizer):
        """
        Slices the LLM output. Handles 1D input_ids (Sequence length only).
        """
        # 1. Find Start Index

        if self.config.vision_end_token:
            end_id = tokenizer.convert_tokens_to_ids(self.config.vision_end_token)
            end_matches = (input_ids == end_id).nonzero(as_tuple=True)
            
            if len(end_matches[0]) > 0:
                text_start_idx = end_matches[0][-1].item() + 1
                vision_end_idx = end_matches[0][-1].item()
                start_id = tokenizer.convert_tokens_to_ids(self.config.image_token_str)
                start_matches = (input_ids == start_id).nonzero(as_tuple=True)
                if len(start_matches[0]) > 0:
                    vision_start_idx = start_matches[0][0].item()
                else:
                    vision_start_idx = max(0, vision_end_idx - self.visual_token_count)

                vision_embeds = final_hidden_states[vision_start_idx : vision_end_idx]
                text_embeds = final_hidden_states[text_start_idx : ]
                
                return vision_embeds, text_embeds

                
        start_token_id = tokenizer.convert_tokens_to_ids(self.config.image_token_str)
        matches = (input_ids == start_token_id).nonzero(as_tuple=True)
        
        # Determine indices based on tensor dimensionality
        # If 1D tensor, matches is (indices,) -> use matches[0]
        # If 2D tensor, matches is (batch_idx, seq_idx) -> use matches[1]
        if len(matches) == 1:
            found_indices = matches[0]
        else:
            found_indices = matches[1]

        # Qwen Fallback: Check for <|vision_start|> explicitly if main check failed
        if len(found_indices) == 0:
            if "<|vision_start|>" in tokenizer.get_vocab():
                 alt_id = tokenizer.convert_tokens_to_ids("<|vision_start|>")
                 matches = (input_ids == alt_id).nonzero(as_tuple=True)
                 if len(matches) == 1: found_indices = matches[0]
                 else: found_indices = matches[1]
        
        if len(found_indices) == 0:
            # If no tokens found, assume Image is at the very start
            print(f"Notice: Image token not found. Assuming index 0.")
            vision_start_idx = 0
        else:
            vision_start_idx = found_indices[0].item()
        
        # 2. Calculate End using Hook Count
        vision_end_idx = vision_start_idx + self.visual_token_count
        
        # Safety Clamp
        if vision_end_idx > final_hidden_states.shape[0]:
            vision_end_idx = final_hidden_states.shape[0]

        # 3. Slice
        vision_embeds = final_hidden_states[vision_start_idx : vision_end_idx]
        text_embeds = final_hidden_states[vision_end_idx : ]
        
        return vision_embeds, text_embeds
```

```python
import os
import glob
import json
import torch
from torch.utils.data import Dataset
from PIL import Image

class VRSBenchDataset(Dataset):
    """
    yeilds:
    - image: PIL Image (RGB)
    - objects: List of dictionaries {'label': str, 'box': [x1, y1, x2, y2]}
    - image_id: Filename for reference
    """
    def __init__(self, annotation_dir, image_dir, max_samples=None):
        self.annotation_dir = annotation_dir
        self.image_dir = image_dir
        self.json_files = glob.glob(os.path.join(annotation_dir, "*.json"))

        if max_samples:
            self.json_files = self.json_files[:max_samples]

        print(f"Found {len(self.json_files)} samples in {annotation_dir}")

    def __len__(self):
        return len(self.json_files)

    def __getitem__(self, idx):
        json_path = self.json_files[idx]

        try:
            with open(json_path, 'r') as f:
                data = json.load(f)
        except Exception as e:
            print(f"Error reading {json_path}: {e}")
            return self.__getitem__((idx + 1) % len(self)) # Skip broken files

        image_filename = data.get('image') or os.path.basename(json_path).replace('.json', '.png')
        image_path = os.path.join(self.image_dir, image_filename)

        if not os.path.exists(image_path):
            if image_path.endswith('.png'):
                image_path = image_path.replace('.png', '.jpg')
            elif image_path.endswith('.jpg'):
                image_path = image_path.replace('.jpg', '.png')

        if not os.path.exists(image_path):
            return self.__getitem__((idx + 1) % len(self))

        try:
            image = Image.open(image_path).convert("RGB")
        except Exception as e:
            print(f"Corrupt image {image_path}: {e}")
            return self.__getitem__((idx + 1) % len(self))

        # Format: {'obj_cls': 'dam', 'obj_coord': [x1, y1, x2, y2]} (Normalized 0-1)
        raw_objects = data.get('objects', [])
        parsed_objects = []
        for obj in raw_objects:
            if 'obj_cls' in obj and 'obj_coord' in obj:
                parsed_objects.append({
                    "label": obj['obj_cls'],
                    "box": obj['obj_coord'] # [x1, y1, x2, y2] normalized
                })

        return {
            "image": image,
            "objects": parsed_objects,
            "image_id": image_filename
        }

class VLMPipeline:
    def __init__(self, processor, config):
        self.processor = processor
        self.config = config

    def process_sample(self, image, caption):
        """
        Prepares inputs for the model.
        Returns: dict of tensors (input_ids, pixel_values, attention_mask)
        """
        model_id = self.config.model_id.lower()

        # Qwen 2.5 VL Logic
        if "qwen" in model_id:
            # Qwen requires specific ChatML formatting
            messages = [
                {
                    "role": "user",
                    "content": [
                        {"type": "image", "image": image},
                        {"type": "text", "text": caption},
                    ],
                }
            ]

            text_prompt = self.processor.apply_chat_template(
                messages, tokenize=False, add_generation_prompt=True
            )

            # padding=True ensures tensor shape consistency if we ever batched (but we use batch=1)
            inputs = self.processor(
                text=[text_prompt],
                images=[image],
                padding=True,
                return_tensors="pt"
            )
            return inputs

        # LLaVA Logic
        elif "llava" in model_id:
            # Standard LLaVA Prompt
            prompt = f"USER: <image>\n{caption}\nASSISTANT:"

            inputs = self.processor(
                text=prompt,
                images=image,
                return_tensors="pt"
            )
            return inputs

        else:
            raise ValueError(f"Pipeline not implemented for {model_id}")
```

```python
import math

class LogitLens:
    def __init__(self, model, processor):
        self.model = model
        self.processor = processor
        self.lm_head = self._find_lm_head()
        
    def _find_lm_head(self):
        candidates = [
            "lm_head",                     # Qwen, Kimi, Standard CausalLM
            "language_model.lm_head",      # LLaVA-Next / Some HF Wrappers
            "model.lm_head"                # Nested cases
        ]
        
        for path in candidates:
            try:
                module = self.model
                for part in path.split('.'):
                    module = getattr(module, part)
                print(f"Found LM Head at: {path}")
                return module
            except AttributeError:
                continue
                
        for name, module in self.model.named_modules():
            if name.endswith("lm_head") and isinstance(module, torch.nn.Linear):
                print(f"Found LM Head at: {name}")
                return module
                
        raise ValueError("Could not locate lm_head in this model.")

    def decode_patches(self, vision_embeds, top_k=3):
        """
        Projects visual embeddings directly to the vocabulary.
        
        Args:
            vision_embeds: Tensor [N_Patches, Hidden_Dim]
        Returns:
            List of Lists: [[(token_str, prob), ...], ...] for each patch
        """
        # 1. Project to Vocab
        # Ensure dtype matches (e.g., float16)
        dtype = self.lm_head.weight.dtype
        vision_embeds = vision_embeds.to(dtype)
        
        with torch.no_grad():
            # [N, Dim] @ [Dim, Vocab] -> [N, Vocab]
            logits = self.lm_head(vision_embeds)
            probs = torch.softmax(logits, dim=-1)
            
            # Get Top K
            topk_probs, topk_ids = torch.topk(probs, top_k, dim=-1)
            
        # 2. Decode to Text
        results = []
        tokenizer = self.processor.tokenizer
        
        for i in range(len(vision_embeds)):
            patch_result = []
            for k in range(top_k):
                token_id = topk_ids[i, k].item()
                prob = topk_probs[i, k].item()
                
                # Decode
                token_str = tokenizer.decode([token_id])
                if not token_str: token_str = "_"
                
                patch_result.append((token_str, prob))
            results.append(patch_result)
            
        return results
        
    def plot_lens(self, image, patch_results, grid_shape=None):
    
        if grid_shape:
            h, w = grid_shape
        else:
            h = int(math.sqrt(len(patch_results)))
            w = h

        fig_size = max(12, w // 2)
        plt.figure(figsize=(fig_size, fig_size * (h/w)))
        plt.imshow(image)
        plt.axis('off')
        
        img_w, img_h = image.size
        step_x = img_w / w
        step_y = img_h / h
        
        count = 0
        for r in range(h):
            for c in range(w):
                if count >= len(patch_results): break
                
                clean_tokens = [str(t) for t in patch_results[count]]
                display_text = "\n".join(clean_tokens)
                
                x = (c * step_x) + (step_x / 2)
                y = (r * step_y) + (step_y / 2)
                
                plt.text(
                    x, y, display_text,
                    color='white', fontsize=5, ha='center', va='center', fontweight='bold',
                    bbox=dict(facecolor='black', alpha=0.4, edgecolor='none', pad=0)
                )
                count += 1
                
        plt.title(f"Logit Lens ({h}x{w} Patches)")
        plt.tight_layout()
        return plt.gcf()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

class VLMVisualizer:
    def __init__(self, output_dir="heatmap_results"):

        self.output_dir = output_dir
        os.makedirs(self.output_dir, exist_ok=True)

    def compute_similarity(self, vision_embeds, text_embeds):
        """
        Returns:
            numpy.ndarray: Shape [N_text_tokens, N_vis_tokens]
        """

        vis_norm = torch.nn.functional.normalize(vision_embeds, p=2, dim=-1)
        text_norm = torch.nn.functional.normalize(text_embeds, p=2, dim=-1)


        # Row i = Text Token i
        # Col j = Vision Token j
        # value = similarity
        # Shape: [N_text, Dim] @ [Dim, N_vis] -> [N_text, N_vis]
        similarity_matrix = torch.matmul(text_norm, vis_norm.T)

        return similarity_matrix.float().cpu().detach().numpy()

    def generate_spatial_heatmap(self, vision_embeds, text_embeds, text_tokens, image_id, object_label, grid_shape=None):
        matrix = self.compute_similarity(vision_embeds, text_embeds)
        
        # Find tokens
        target_indices = []
        normalized_label = object_label.lower().strip()
        for idx, token in enumerate(text_tokens):
            t_clean = str(token).lower().strip().replace('ġ', '').replace('</w>', '')
            if normalized_label in t_clean or t_clean in normalized_label:
                if len(t_clean) > 2 or t_clean == normalized_label:
                    target_indices.append(idx)
        if not target_indices: target_indices = [-1]

        # Average scores
        target_row = matrix[target_indices, :].mean(axis=0)

        n_tokens = target_row.shape[0]
        
        if grid_shape:
            h, w = grid_shape
            if h * w != n_tokens:
                print(f"⚠️ Grid Mismatch: {h}x{w}={h*w} vs {n_tokens} tokens. Fallback to sqrt.")
                side = int(math.sqrt(n_tokens))
                h, w = side, side
        else:
            h = int(math.sqrt(n_tokens))
            w = h

        valid_tokens = h * w
        spatial_map = target_row[:valid_tokens].reshape(h, w)
        
        # Plot
        plt.figure(figsize=(w/3, h/3)) # Aspect ratio preserved
        spatial_map = (spatial_map - spatial_map.min()) / (spatial_map.max() - spatial_map.min() + 1e-8)
        
        sns.heatmap(
            spatial_map,
            cmap="jet",
            square=True,
            xticklabels=False,
            yticklabels=False,
            cbar=True,
            cbar_kws={'label': 'Attention'}
        )
        
        plt.title(f"Locating: '{object_label}'\n({h}x{w} Grid)")
        clean_id = str(image_id).replace("/", "_").replace(".", "_")
        filename = f"{self.output_dir}/{clean_id}_{object_label}_spatial.png"
        
        plt.tight_layout()
        plt.savefig(filename, dpi=150, bbox_inches='tight')
        plt.close()
        
        return filename
'''
    def generate_heatmap(self, vision_embeds, text_embeds, tokens, image_id, object_label):

        matrix = self.compute_similarity(vision_embeds, text_embeds)
        width = min(20, max(10, matrix.shape[1] / 50))
        height = max(6, matrix.shape[0] / 2)

        plt.figure(figsize=(width, height))

        v_min = matrix.min()
        v_max = matrix.max()
        ax = sns.heatmap(
            matrix,
            yticklabels=tokens,
            xticklabels=False,
            cmap="viridis",
            cbar_kws={'label': 'Cosine Similarity'},
            vmin=v_min,
            vmax=v_max  
        )

        plt.title(f"Cross-Modal Alignment: '{object_label}'\nImage ID: {image_id}")
        plt.xlabel(f"Visual Tokens (Flattened Sequence: 0 to {matrix.shape[1]})")
        plt.ylabel("Text Prompt Tokens")


        clean_id = str(image_id).replace("/", "_").replace(".", "_")
        filename = f"{self.output_dir}/{clean_id}_{object_label}.png"

        plt.tight_layout()
        plt.savefig(filename, dpi=150)
        plt.close()
        return filename

'''
```

```python
import traceback
from transformers import (
    AutoProcessor,
    AutoModelForCausalLM,
    LlavaForConditionalGeneration,
    Qwen2VLForConditionalGeneration
)

from transformers.utils import is_accelerate_available

def main(MODEL_KEY):
    # Options: "qwen2.5-vl-3b", "llava-1.5", "kimi-vl"
    #MODEL_KEY = "qwen2.5-vl-3b"

    #ANNOTATION_DIR = "./VRSBench_val/Annotations_val"
    #IMAGE_DIR = "./VRSBench_val/Images_val"
    OUTPUT_DIR = f"./heatmap_results_{MODEL_KEY}"

    MAX_SAMPLES = 5  # can be increased

    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"Starting Evaluation for [{MODEL_KEY}] on {device}...")

    if MODEL_KEY not in SUPPORTED_MODELS:
        raise ValueError(f"Model {MODEL_KEY} not found in vlm_config.py")
    config = SUPPORTED_MODELS[MODEL_KEY]


    print(f"   Loading model: {config.model_id}")

    # For Kimi
    if config.use_causal_lm:
        print("   -> Mode: AutoModelForCausalLM (Custom/Kimi)")
        print(device)
        model = AutoModelForCausalLM.from_pretrained(
            config.model_id,
            torch_dtype=torch.float16 if device=="cuda" else torch.float32,
            device_map="auto",
            trust_remote_code=True
        )

    # Qwen
    elif "qwen" in MODEL_KEY:
        print("   -> Mode: Qwen2VLForConditionalGeneration")
        model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
            config.model_id,
            torch_dtype=torch.float16 if device=="cuda" else torch.float32,
            device_map="auto"
        )

    # LLaVA
    else:
        print("   -> Mode: LlavaForConditionalGeneration")
        
        # The A5000 works best with device_map="auto" (requires the 'accelerate' fix above)
        model = LlavaForConditionalGeneration.from_pretrained(
            config.model_id, 
            torch_dtype=torch.float16, 
            device_map="auto" 
        )

    processor = AutoProcessor.from_pretrained(config.model_id, trust_remote_code=True)


    pipeline = VLMPipeline(processor, config)
    tracker = VLMTracker(model, config)
    visualizer = VLMVisualizer(output_dir=OUTPUT_DIR)

    if os.path.exists(LOCAL_ANNOTATION_DIR):
        dataset = VRSBenchDataset(LOCAL_ANNOTATION_DIR, LOCAL_IMAGE_DIR, max_samples=MAX_SAMPLES)
        print(f"Dataset Loaded: {len(dataset)} samples found.")
    else:
        print("Data directories not found. Using Dummy Data for test?")
        return


    print("Starting Inference")
    logit_lens = LogitLens(model, processor)
    

    for i, sample in enumerate(dataset):
        image = sample["image"]
        image_id = sample["image_id"]
        objects = sample["objects"]

        if not objects:
            continue


        # We pick the first object and create a caption like "This is a [label]."
        # We want to see if the tokens for "[label]" attend to the correct image patch.
        target_obj = objects[0]
        label = target_obj["label"]
        caption = f"This image contains a {label}."

        print(f"[{i+1}/{len(dataset)}] ID: {image_id} | Target: '{label}'")

        try:
            inputs = pipeline.process_sample(image, caption).to(device)


            # The 'track()' context manager captures the exact visual token count
            with tracker.track():
                with torch.autocast(device_type="cuda", dtype=torch.float16):
                    outputs = model(**inputs, output_hidden_states=True)


            # Get last layer hidden states (Batch, Seq, Dim) -> Squeeze batch
            last_hidden = outputs.hidden_states[-1].squeeze(0)

            # Retrieve input_ids to find where text starts/ends
            input_ids = inputs.input_ids[0]

            vis_embeds, text_embeds = tracker.split_embeddings(
                last_hidden,
                input_ids,
                processor.tokenizer
            )


            text_len = text_embeds.shape[0]
            text_ids = input_ids[-text_len:]
            
            raw_tokens = processor.tokenizer.convert_ids_to_tokens(text_ids)
            decoded_tokens = []
            for t in raw_tokens:
                if isinstance(t, bytes):
                    t = t.decode('utf-8', errors='replace')
                decoded_tokens.append(t)


            patch_tokens = logit_lens.decode_patches(vis_embeds, top_k=3)
            fig = logit_lens.plot_lens(image, patch_tokens, grid_size=24)
                

            save_path = visualizer.generate_spatial_heatmap(
                vis_embeds, 
                text_embeds, 
                decoded_tokens, 
                image_id, 
                label
            )
            print(f"Spatial Heatmap saved: {save_path}")

            lens_filename = f"{OUTPUT_DIR}/{image_id}_logit_lens_top3.png"
            fig.savefig(lens_filename, dpi=200)
            plt.close(fig)

            print("Logit Lens saved")

        except Exception as e:
            print(f"Failed: {e}")
            import traceback
            traceback.print_exc()

    print(f"\nEvaluation Complete. Results in {OUTPUT_DIR}")
```
